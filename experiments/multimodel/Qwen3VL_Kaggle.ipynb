# Qwen3-VL-30B-A3B-Instruct: experiments 1 and 2

Attach the uploaded `kaggle_new_models` dataset, enable Internet, and select **T4 x2** manually. This notebook runs only `qwen3vl`. Default precision is expert-NF4; TRC uses eight separately encoded frames. See README.md for methodology and limits.


```python
from pathlib import Path
import subprocess, sys, os, shutil
matches = list(Path('/kaggle/input').rglob('kaggle_new_models/run.py'))
assert len(matches) == 1, f'Expected one uploaded package; found {matches}'
CODE = matches[0].parent
MODEL = 'qwen3vl'
OUTPUT = Path('/kaggle/working/new_model_results')
os.environ['HF_XET_CHUNK_CACHE_SIZE_BYTES'] = '0'
subprocess.run(['bash', str(CODE / 'setup_kaggle.sh')], check=True)
subprocess.run([sys.executable, '-m', 'pytest', str(CODE), '-q', '-p', 'no:cacheprovider'], check=True)
```

## Smoke test
Two examples per experiment, in separate folders. Weight downloads are still full size, but only one shard is retained at a time. A failed native-equivalence check or incomplete matrix stops the subprocess.


```python
subprocess.run([sys.executable, str(CODE / 'run.py'), '--model', MODEL, '--smoke'], check=True)
subprocess.run([sys.executable, str(CODE / 'analyze.py'), '--model', MODEL, '--smoke'], check=True)
```

```python
import json, pandas as pd
for experiment in (1, 2):
    root = OUTPUT / MODEL / f'experiment_{experiment}_smoke'
    print(f'Experiment {experiment}:', json.loads((root / 'native_equivalence.json').read_text()))
    display(pd.read_csv(root / 'summaries' / 'summary.csv'))
```

## Full runs

Experiment 2 runs all four frame conditions. Rerun the same cell to resume; results must still be present under `/kaggle/working`. Model loading downloads shards again because the smoke subprocess released its model. Full runtime has not been measured; save outputs before the Kaggle session expires.


```python
subprocess.run([sys.executable, str(CODE / 'run.py'), '--model', MODEL, '--limit', '256'], check=True)
subprocess.run([sys.executable, str(CODE / 'analyze.py'), '--model', MODEL, '--limit', '256'], check=True)
```

```python
for experiment in (1, 2):
    root = OUTPUT / MODEL / f'experiment_{experiment}_n256'
    display(pd.read_csv(root / 'summaries' / 'summary.csv'))
display(pd.read_csv(OUTPUT / MODEL / 'experiment_2_n256/summaries/condition_accuracy.csv'))
```
